```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/employee_data_cleaning.csv")
df
```

```python
print("Shape of Dataset:", df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(8,5))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.title("Missing Values Heatmap")
plt.show()
```

```python
df["Salary"] = pd.to_numeric(df["Salary"], errors='coerce')

df["Salary"].fillna(df["Salary"].mean(), inplace=True)
df["Experience"].fillna(df["Experience"].median(), inplace=True)
df["Emp_Name"].fillna("Unknown", inplace=True)
df["Location"].fillna("Not Specified", inplace=True)

df
```

```python
duplicates = df[df.duplicated()]

print("Duplicate Records:")
print(duplicates)
```

```python
df = df.drop_duplicates()

print("Shape After Removing Duplicates:", df.shape)

df
```

```python
df["Salary"] = pd.to_numeric(df["Salary"], errors='coerce')

invalid_rows = df[df["Salary"].isna()]

print("Rows with Invalid Salary Values:")
print(invalid_rows)
```

```python
df["Salary"] = pd.to_numeric(df["Salary"], errors='coerce')

df
```

```python
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df["Salary"] < lower_bound) |
              (df["Salary"] > upper_bound)]

print("Outliers Detected:")
print(outliers)
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(y=df["Salary"])
plt.title("Salary Distribution Before Removing Outliers")
plt.show()
```

```python
df_no_outliers = df[
    (df["Salary"] >= lower_bound) &
    (df["Salary"] <= upper_bound)
]

df_no_outliers
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(y=df_no_outliers["Salary"])
plt.title("Salary Distribution After Removing Outliers")
plt.show()
```

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
sns.boxplot(y=df["Salary"])
plt.title("Before Outlier Removal")

plt.subplot(1,2,2)
sns.boxplot(y=df_no_outliers["Salary"])
plt.title("After Outlier Removal")

plt.tight_layout()
plt.show()
```

```python
      df_no_outliers.to_csv("cleaned_employee_data.csv", index=False)

print("Cleaned Dataset Saved Successfully")
```
