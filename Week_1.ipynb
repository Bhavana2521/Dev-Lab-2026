```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Student":["A","B","C","D"],
    "Marks":[85,90,78,88]
}

df = pd.DataFrame(data)

df.plot(
    x="Student",
    y="Marks",
    kind="bar"
)

plt.title("Student Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ["Python","Java","C++","SQL"]
students = [40,30,20,10]

plt.pie(
    students,
    labels=subjects,
    autopct='%1.1f%%'
)

plt.title("Programming Language Usage")

plt.show()
```
