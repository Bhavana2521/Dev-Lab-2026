```python
import pandas as pd

data = {
    'Age': [20, 30, 40, 40, 50],
    'Salary': [50000, 60000, 70000, 55000, 80000]
}

df = pd.DataFrame(data)

print(df.describe())
```

```python
#mean,median and mode

print("Mean:",df['Age'].mean())
print("Median:",df['Age'].median())
print("Mode:",df['Age'].mode()[0])
```

```python
#variance and standard deviation
print("Variance:", df['Age'].var())
print("Standard Deviation:", df['Age'].std())
```

```python
#count,max,min
print("count:",df['Age'].count())
print("max:",df['Age'].max())
print("min:",df['Age'].min())
```

```python
print("summary report")
print(df.shape)
print(df.columns)
print(df.dtypes)
print(df.isnull().sum())
print(df.describe())
print("\nAge Statistics")
print("count:", df['Age'].count())
print("Mean:", df['Age'].mean())
print("Median:", df['Age'].median())
print("Mode:", df['Age'].mode()[0])
print("Variance:", df['Age'].var())
print("Standard Deviation:", df['Age'].std())
print('maximum age:',df['Age'].max())
print('minimum age:',df['Age'].min())
print("\nSalary Statistics")
print("Mean:", df['Salary'].mean())
print("Median:", df['Salary'].median())
print("mode:", df['Salary'].mode()[0])
print("var:", df['Salary'].var())
print("standard deviation:", df['Salary'].std())
print("Maximum Salary:", df['Salary'].max())
print("Minimum Salary:", df['Salary'].min())
```

```python
#numeric dataset mean,median,mode generated test case
stats=df.describe()
assert not stats.empty,"stats are not generated"
print("pass:stats generated")
print(stats)
```

```python
#mixed data test case
import pandas as pd

df = pd.DataFrame({
    'Name': ['bhavana', 'Priya', 'sony'],
    'Age': [25, 30, 35],
    'City': ['Delhi', 'Mumbai', 'Chennai']
})

summary = df.describe(include='all') #Without include='all', Pandas summarizes only numeric columns.

assert not summary.empty, "Summary not generated"

print("PASS: Summary Report Generated")
print(summary)
```

```python
#large data set test case
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Salary': np.random.randint(20000,100000,100000)   #lower limit included,upper limit exclucded,generate 100000 random numbers
})

stats = df.describe()

assert not stats.empty, "Statistics generation failed"

print("PASS: Large Dataset Statistics Computed")
print(stats)
```

```python
#non numerical data set partial stats test case
import pandas as pd

df = pd.DataFrame({
    'Name': ['mahi', 'dhoni', 'chiru']
})

try:
    stats = df.describe()

    print("PASS: Partial Statistics Generated")
    print(stats)
#top --->Most frequently occurring value   #freq-->Frequency (count) of the top value
except Exception as e:
    print("FAIL:", e)
```

```python
#empty data set test case
import pandas as pd

df = pd.DataFrame()

try:
    assert not df.empty,"Dataset is empty"
    print(df.describe())

except AssertionError as e:
    print("FAIL:", e)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Age': [20, 30, 40, 40, 50],
    'Salary': [50000, 60000, 70000, 55000, 80000]
}

df = pd.DataFrame(data)

plt.bar(range(len(df)), df['Age'])
plt.title("Age Distribution")
plt.xlabel("Person")
plt.ylabel("Age")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Age': [20, 30, 40, 40, 50]
}

df = pd.DataFrame(data)

plt.boxplot(df['Age'])

plt.title("Box Plot of Age")
plt.ylabel("Age")
plt.show()
```
