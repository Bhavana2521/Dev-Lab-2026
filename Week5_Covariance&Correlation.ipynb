```python
import pandas as pd

df = pd.DataFrame({
    'Study_Hours': [1, 2, 3, 4, 5],
    'Marks': [40, 50, 65, 70, 75]
})

print("Dataset")
print(df)

print("\nCovariance Matrix")
print(df.cov())

print("\nCorrelation Matrix")
print(df.corr())

print("\nCovariance (Hours vs Marks):")
print(df['Study_Hours'].cov(df['Marks']))

print("\nCorrelation (Hours vs Marks):")
print(df['Study_Hours'].corr(df['Marks']))
```

```python
#NUMERIC FEATURES FOR CORRELATION MATRIX
import pandas as pd

df = pd.DataFrame({
    'Study_Hours': [1, 2, 3, 4, 5],
    'Marks': [40, 50, 60, 70, 80]
})

corr_matrix = df.corr()
assert not corr_matrix.empty,"Correlation matrix not generated"

print("PASS: Correlation Matrix Generated")
print(corr_matrix)
```

```python
#Positive Relation
import pandas as pd

df = pd.DataFrame({
    'X': [1, 2, 3, 4, 5],
    'Y': [10, 20, 30, 40, 50]
})

corr = df['X'].corr(df['Y'])

assert corr > 0.9,"Positive correlation not detected"

print("PASS: High Positive Correlation")
print("Correlation =", corr)
```

```python
#Negative Relation
import pandas as pd

df = pd.DataFrame({
    'X': [1, 2, 3, 4, 5],
    'Y': [50, 40, 30, 20, 10]
})

corr = df['X'].corr(df['Y'])

assert corr < 0, "Negative correlation not detected"

print("PASS: Negative Correlation")
print("Correlation =", corr)
```

```python
#Constant Column (one column has same value)
import pandas as pd

df = pd.DataFrame({
    'X': [10, 10, 10, 10, 10],
    'Y': [1, 2, 3, 4, 5]
})

corr = df['X'].corr(df['Y'])

assert pd.isna(corr),"Expected NaN for constant column"

print("PASS: NaN Correlation Detected")
print("Correlation =", corr)
```

```python
#Non-Numeric Data test case
import pandas as pd

df = pd.DataFrame({
    'Name': ['Bhavana', 'priya', 'sony'],
    'City': ['Delhi', 'Mumbai', 'Chennai']
})

try:
    corr = df['Name'].corr(df['City'])

    print("FAIL: Correlation should not be computed")

except Exception as e:
    print("PASS: Non-Numeric Data Detected")
    print("Error:", e)
```
